### Group By 'text' Column

- **What:** Count the occurrences of each patient/recording description
- **Why:** Understand the distribution of your dataset (e.g., how many recordings per patient type, AF vs non-AF cases, etc.).

```python
# Create unique patient identifier based on age, gender, BMI
df['patient_id'] = df.groupby(['text']).ngroup().add(1).apply(lambda x: f'P{x}')

# Or more explicitly extract features:
import re

def extract_patient_info(text):
    age = re.search(r'(\d+)-year-old', text).group(1)
    gender = re.search(r'(male|female) patient', text).group(1)
    bmi = re.search(r'BMI (\d+\.\d+)', text).group(1)
    return f"{age}_{gender}_{bmi}"

df['patient_key'] = df['text'].apply(extract_patient_info)
df['patient_id'] = pd.factorize(df['patient_key'])[0] + 1
df['patient_id'] = df['patient_id'].apply(lambda x: f'P{x}')

df.groupby(['patient_id', 'text']).size()
```
