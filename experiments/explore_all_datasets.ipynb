### Initial note
For our purposes, we want **raw continuous time series** suitable for resampling, filtering, DC-offset removal, windowing, and normalization. In the PulseLM-formatted datasets, **none of the exported signals are raw continuous time series**; they are already **segmented and task-formatted**, so they should not be used directly for the full preprocessing pipeline.
