```python
from datasets import load_dataset, get_dataset_config_names, concatenate_datasets

dataset_names = get_dataset_config_names("Manhph2211/PulseLM")
dataset_names
```

```python
len(dataset_names)
```

```python
for name in dataset_names:
    print(f"Processing: {name}")
    
    # Load dataset
    dataset = load_dataset("Manhph2211/PulseLM", name, split="train").select_columns(["signal", "text", "qa"])
    
    # Save as parquet
    filename = f"../data/pulselm/{name}.parquet"
    dataset.to_parquet(filename)
    
    print(f"  Saved {len(dataset):,} samples to {filename}")
    print()

print("All datasets saved!")
```
